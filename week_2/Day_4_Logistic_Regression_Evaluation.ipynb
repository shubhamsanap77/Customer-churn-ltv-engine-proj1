```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath('.'))
from data_loader import load_raw_data
from feature_engineering import FeatureEngineer
from train_logistic_regression import LogisticRegressionPipeline

%matplotlib inline
```

## 1. Load Data and Feature Engineering

```python
df = load_raw_data()
fe = FeatureEngineer()
X_train, X_test, y_train, y_test, features = fe.fit_transform(df)

print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
print(f"Churn rate in training set: {y_train.mean():.2%}")
```

## 2. Train Logistic Regression & Evaluate Metrics

```python
pipeline = LogisticRegressionPipeline()
results_df = pipeline.train_and_evaluate(X_train, X_test, y_train, y_test)
pipeline.generate_plots(y_test)
pipeline.save_artifacts(features, fe.scaler)

results_df
```

## 4. Key Takeaways for Day 4
- **Class-Balanced Logistic Regression** improves Churn **Recall from 54.0% to 79.7%** and **F1-Score from 0.599 to 0.613**.
