```python
# ============================================================
# WEEK 4 - DAY 1
# LTV CALCULATION & CUSTOMER VALUE FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
import os

print("=" * 70)
print("WEEK 4 - DAY 1")
print("LTV CALCULATION & CUSTOMER VALUE FEATURE ENGINEERING")
print("=" * 70)


# ============================================================
# 1. LOAD DATASET
# ============================================================

data_path = "../../../data/initial_engineered_customer_dataset.csv"

df = pd.read_csv(data_path)

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)


# ============================================================
# 2. BASIC DATA VALIDATION
# ============================================================

print("\n--- DATA VALIDATION ---")

print("Total Customers:", len(df))
print("Duplicate Rows:", df.duplicated().sum())

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 3. CLEAN TOTAL CHARGES
# ============================================================

print("\n--- CLEANING TOTALCHARGES ---")

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

missing_totalcharges = df["TotalCharges"].isnull().sum()

print("Missing TotalCharges before cleaning:",
      missing_totalcharges)

df["TotalCharges"] = df["TotalCharges"].fillna(
    df["TotalCharges"].median()
)

print("Missing TotalCharges after cleaning:",
      df["TotalCharges"].isnull().sum())


# ============================================================
# 4. SELECT ACTIVE CUSTOMERS
# ============================================================

print("\n--- ACTIVE CUSTOMERS ---")

active_customers = df[
    df["Churn"] == "No"
].copy()

print("Total Customers:", len(df))
print("Active Customers:", len(active_customers))


# ============================================================
# 5. EXPECTED REMAINING MONTHS
# ============================================================

print("\n--- EXPECTED REMAINING MONTHS ---")

# Maximum tenure assumption = 72 months

active_customers["ExpectedRemainingMonths"] = (
    72 - active_customers["tenure"]
)

# Minimum remaining lifetime = 1 month

active_customers["ExpectedRemainingMonths"] = (
    active_customers["ExpectedRemainingMonths"]
    .clip(lower=1)
)

print(
    active_customers[
        ["tenure", "ExpectedRemainingMonths"]
    ].head()
)


# ============================================================
# 6. ESTIMATED LTV CALCULATION
# ============================================================

print("\n--- ESTIMATED LTV ---")

active_customers["Estimated_LTV"] = (
    active_customers["MonthlyCharges"]
    * active_customers["ExpectedRemainingMonths"]
)

print("Estimated LTV calculated successfully.")

print("\nLTV Statistics:")
print(
    active_customers["Estimated_LTV"].describe()
)


# ============================================================
# 7. CUSTOMER VALUE FEATURE ENGINEERING
# ============================================================

print("\n--- CUSTOMER VALUE FEATURES ---")


# Historical charges per month of tenure

active_customers["TotalChargesPerTenure"] = (
    active_customers["TotalCharges"]
    / active_customers["tenure"].replace(0, np.nan)
)

active_customers["TotalChargesPerTenure"] = (
    active_customers["TotalChargesPerTenure"]
    .fillna(0)
)


# Monthly charges relative to tenure

active_customers["MonthlyChargePerTenure"] = (
    active_customers["MonthlyCharges"]
    / active_customers["tenure"].replace(0, np.nan)
)

active_customers["MonthlyChargePerTenure"] = (
    active_customers["MonthlyChargePerTenure"]
    .fillna(0)
)


print("Customer value features created successfully.")


# ============================================================
# 8. DISPLAY CUSTOMER VALUE FEATURES
# ============================================================

value_features = [
    "customerID",
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "ExpectedRemainingMonths",
    "TotalChargesPerTenure",
    "MonthlyChargePerTenure",
    "Estimated_LTV"
]

print("\nCustomer Value Features:")
print(
    active_customers[value_features].head(10)
)


# ============================================================
# 9. CUSTOMER VALUE SUMMARY
# ============================================================

print("\n--- CUSTOMER VALUE SUMMARY ---")

summary_features = [
    "MonthlyCharges",
    "TotalCharges",
    "ExpectedRemainingMonths",
    "TotalChargesPerTenure",
    "MonthlyChargePerTenure",
    "Estimated_LTV"
]

print(
    active_customers[summary_features].describe()
)


# ============================================================
# 10. SAVE FINAL DAY 1 DATASET
# ============================================================

print("\n--- SAVING FINAL DATASET ---")

output_directory = "../../../data"

output_file = (
    output_directory
    + "/week4_day1_ltv_customer_value.csv"
)

active_customers.to_csv(
    output_file,
    index=False
)

print("Final dataset saved successfully!")
print("Saved at:", output_file)


# ============================================================
# 11. FINAL VERIFICATION
# ============================================================

print("\n--- FINAL VERIFICATION ---")

final_df = pd.read_csv(output_file)

print("Final Dataset Shape:", final_df.shape)

print("Total Missing Values:",
      final_df.isnull().sum().sum())

print("Duplicate Rows:",
      final_df.duplicated().sum())

print("\nNew Features Added:")

print("1. ExpectedRemainingMonths")
print("2. Estimated_LTV")
print("3. TotalChargesPerTenure")
print("4. MonthlyChargePerTenure")


# ============================================================
# COMPLETION
# ============================================================

print("\n" + "=" * 70)
print("WEEK 4 - DAY 1 COMPLETED SUCCESSFULLY")
print("=" * 70)
```
